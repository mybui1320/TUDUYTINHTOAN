```python
s= list(map(int,input().split()))
k=int(input())

res=[]
for i in range(1, len(s)):

    for j in range(i+1, len(s)):
        if s[i]+s[j]==k:
            res.append((s[i],s[j]))

res.sort()
print(res)
```

[Lọc dictionary theo giá trị]
Nhập vào một dictionary có key là chuỗi và value là số nguyên, và một số
nguyên k ở dòng tiếp theo. Hãy tạo một dictionary mới chỉ chứa các cặp keyvalue từ dictionary ban đầu mà có value lớn hơn k. In ra dictionary mới.

```python

a= input().split()
k = int(input())
d={}
for i in range(0,len(a)-1,2):
    d[a[i]]=int(a[i+1])
res = {}

for key, value in d.items():
    if value > k:
        res[key] = value

print(res)
```

```python
def min_and_pos(lst):
  d = {}
  
  for x in lst:
      d[x] = d.get(x, 0) + 1
  
  max_count = max(d.values())
  
  res = []
  
  for k in d:
      if d[k] == max_count:
          res.append(k)
  return min(res)

lst= list(map(int, input().split()))    
print(min_and_pos(lst))

try:
  a = list(map(int, input().split()))
  b = list(map(int, input().split()))
  s=[]
  for x in a:
    if x in b and x not in s:
      s.append(x)

  print(s)
except EOFError:
  print([])
```

```python
def fib(n):
    # add your code here
    if n == 0:
        return 0

    if n == 1:
        return 1

    a = 0
    b = 1

    for i in range(2, n + 1):
        c = a + b
        a = b
        b = c

    return b

n = int(input().strip())
print(fib(n))
```
