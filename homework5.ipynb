W5A18 VIết hàm kiểm tra một số có phải số đối xứng không, xử lý trên dữ liệu kiểu số nguyên.


```python
def reverse(a):
    length=len(str(a))
    so_cuoi = 0
    reversed_num = 0
    for _ in range(length):
        so_cuoi = a%10
        a = a//10
        reversed_num = reversed_num*10 + so_cuoi
    return reversed_num
def check_palindrome(a):
    if a == reverse(a):
        return True
    else:
        return False
```

Viết hàm sum_1_to_n(n) trả về tổng các số x từ 1 đến n, nếu x có các chữ số đôi một khác nhau
Nếu n ≤ 0 trả về 0.

cho 1 chuỗi chỉ gồm các ký tự chữ cái và chữ số
tính tổng các số có mặt trong chuỗi
vd 123hbjn699 => 123+699

tính trung bình các số có mặt trong chuỗi, lấy đến 2 chữ số phần thật phân
vd 1sfv0vbj6jkn3 => (1+0+6+3)/4

```python
s= input().strip()
number=0
tong=0
for c in s:
    if '0'<=c<='9':
        number= number*10 +(ord(c)- ord(0))
    else:
        tong+=number


def trung_binh(s):
    cnt=0
    tong=0
    for c in s:
        if c.isdigit(c):
            cnt+=1
            tong+=int(c)

    tb= tong/cnt
    return (f"{tb:.2f}")

print(trung_binh(s))
    
```
