Sử dụng các toán tử bitwise để xác định xem một số có phải là lũy thừa của 2 hay không.
Gợi ý: Đối với lũy thừa của 2, n & (n-1) == 0 


```python
n= int(input())
if ((n>0) and (n &(n-1)==0)):
    print(f"{n} là lũy thừa của 2")
else:
    print(f"{n} không phải là lũy thừa của 2")
```

[Xếp loại học lực]
Nhập vào điểm trung bình của một học sinh và in ra học lực của học sinh đó. Xếp loại học lực theo quy tắc:
- >= 8.0: Giỏi
- >= 6.5: Khá
- >= 5.0: Trung bình
- < 5.0: Yếu

```python
điểm = float(input("Nhập điểm từ 0-10:"))
if 8 <= điểm <= 10:
    print("Học lực: Giỏi")
elif 6.5<= điểm < 8:
    print("Học lực: Khá")
elif 5 <= điểm < 6.5:
    print("Học lực: Trung bình")
elif 0<= điểm < 5:
    print("Học lực: Yếu")
```
